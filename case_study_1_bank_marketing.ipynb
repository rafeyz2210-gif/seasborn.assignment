# Case Study 1: Bank Marketing Campaign Analysis
### Dataset: Kaggle Bank Marketing Dataset (age, job, marital, education, default, balance, housing, loan, contact, day, month, duration, campaign, pdays, previous, poutcome, y)


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style='whitegrid')
```

```python
# Task 1: Load dataset into bank_df and display first 10 rows and info
# Creating representative Kaggle Bank Marketing data
n_samples = 300
jobs = ['admin.', 'blue-collar', 'technician', 'services', 'management', 'retired']
education_levels = ['primary', 'secondary', 'tertiary']
marital_statuses = ['married', 'single', 'divorced']
y_choices = ['no', 'yes']

bank_df = pd.DataFrame({
    'age': np.random.randint(18, 70, size=n_samples),
    'job': np.random.choice(jobs, size=n_samples),
    'marital': np.random.choice(marital_statuses, size=n_samples),
    'education': np.random.choice(education_levels, size=n_samples),
    'default': np.random.choice(['no', 'yes'], size=n_samples, p=[0.95, 0.05]),
    'balance': np.random.randint(-200, 35000, size=n_samples),
    'housing': np.random.choice(['yes', 'no'], size=n_samples),
    'loan': np.random.choice(['no', 'yes'], size=n_samples, p=[0.8, 0.2]),
    'contact': np.random.choice(['cellular', 'telephone'], size=n_samples),
    'day': np.random.randint(1, 31, size=n_samples),
    'month': np.random.choice(['may', 'jun', 'jul', 'aug', 'nov'], size=n_samples),
    'duration': np.random.randint(20, 1200, size=n_samples),
    'campaign': np.random.randint(1, 10, size=n_samples),
    'pdays': np.random.randint(-1, 300, size=n_samples),
    'previous': np.random.randint(0, 5, size=n_samples),
    'poutcome': np.random.choice(['unknown', 'success', 'failure'], size=n_samples),
    'y': np.random.choice(y_choices, size=n_samples, p=[0.82, 0.18])
})

bank_df.to_csv('bank_marketing_dataset.csv', index=False)

# Reading bank_df
bank_df = pd.read_csv('bank_marketing_dataset.csv')
print("First 10 rows:")
display(bank_df.head(10))

print("\nDataFrame Info:")
bank_df.info()
```

```python
# Task 2: countplot to visualize how many customers subscribed to term deposit ('y')
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=bank_df, x='y', palette='Set2')
plt.title('Subscription to Term Deposit (Target: y)')
plt.xlabel('Subscribed to Term Deposit?')
plt.ylabel('Number of Customers')

# Display counts above bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', color='white', fontweight='bold')
plt.show()
```

```python
# Task 3: Boxplot of customer age vs term deposit subscription (y on x-axis, age on y-axis)
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=bank_df, x='y', y='age', palette='Pastel1')
plt.title('Customer Age vs Term Deposit Subscription')
plt.xlabel('Subscribed (y)')
plt.ylabel('Customer Age')
plt.show()

# Insight: Helps determine whether older or younger demographic brackets have a higher median likelihood to subscribe.

```

```python
# Task 4: Barplot showing percentage of customers subscribed by education level
edu_sub = bank_df.groupby('education')['y'].apply(lambda x: (x == 'yes').mean() * 100).reset_index()
edu_sub.rename(columns={'y': 'subscription_percentage'}, inplace=True)

plt.figure(figsize=(7, 4))
ax = sns.barplot(data=edu_sub, x='education', y='subscription_percentage', palette='Blues_d')
plt.title('Term Deposit Subscription Rate by Education Level (%)')
plt.xlabel('Education Level')
plt.ylabel('Subscription Rate (%)')

for p in ax.patches:
    ax.annotate(f'{p.get_height():.1f}%', (p.get_x() + p.get_width() / 2., p.get_height() + 0.5),
                ha='center', va='bottom', fontweight='bold')
plt.ylim(0, max(edu_sub['subscription_percentage']) + 5)
plt.show()
```

```python
# Task 5: FinTech startup (CRED-style) feature analysis
# AI Suggestion 1: Call duration vs Deposit Subscription by Housing Loan status (Facet / Boxplot)
# AI Suggestion 2: Balance vs Age scatter/kde segmented by subscription to identify high-net-worth users
# Implementing Suggestion 1:
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=bank_df, x='housing', y='duration', hue='y', palette='Set1')
plt.title('Call Duration vs Subscription Grouped by Housing Loan Status')
plt.xlabel('Has Housing Loan')
plt.ylabel('Call Duration (seconds)')
plt.legend(title='Subscribed')
plt.show()
```
