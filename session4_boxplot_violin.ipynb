# Session 4: Boxplot & Violin Plot (Outlier Visualization)
### Topics: sns.boxplot(), sns.violinplot(), Detecting outliers


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style="whitegrid")
```

```python
# Task 1: Load Flipkart product prices and use boxplot()
flipkart_prices = pd.DataFrame({
    'product_id': range(1, 36),
    'price': list(np.random.normal(loc=1200, scale=300, size=30)) + [3500, 4200, 3900, 150, 200]
})
flipkart_prices.to_csv('flipkart_prices.csv', index=False)

df_flipkart = pd.read_csv('flipkart_prices.csv')

plt.figure(figsize=(7, 3.5))
sns.boxplot(x=df_flipkart['price'], color='lightgreen')
plt.title('Flipkart Product Price Distribution')
plt.xlabel('Price (₹)')
plt.show()
```

```python
# Task 2: violinplot() comparing order amounts for two Zomato categories
zomato_orders = pd.DataFrame({
    'category': ['Fast Food'] * 40 + ['Desserts'] * 40,
    'order_amount': list(np.random.normal(loc=350, scale=80, size=40)) + list(np.random.normal(loc=220, scale=45, size=40))
})

plt.figure(figsize=(7, 4))
sns.violinplot(data=zomato_orders, x='category', y='order_amount', palette='Pastel1')
plt.title('Zomato Order Amounts: Fast Food vs Desserts')
plt.xlabel('Food Category')
plt.ylabel('Order Amount (₹)')
plt.show()
```

```python
# Task 3: IPL ticket prices boxplot and identifying outliers
ipl_ticket_prices = np.array([400, 450, 500, 500, 550, 600, 650, 700, 750, 800, 850, 900, 950, 2500, 3200])

plt.figure(figsize=(7, 3))
sns.boxplot(x=ipl_ticket_prices, color='skyblue')
plt.title('IPL Match Ticket Prices Boxplot')
plt.xlabel('Ticket Price (₹)')
plt.show()

# Calculate IQR to identify outliers programmatically
q1 = np.percentile(ipl_ticket_prices, 25)
q3 = np.percentile(ipl_ticket_prices, 75)
iqr = q3 - q1
upper_whisker = q3 + 1.5 * iqr
lower_whisker = q1 - 1.5 * iqr

outliers = ipl_ticket_prices[(ipl_ticket_prices < lower_whisker) | (ipl_ticket_prices > upper_whisker)]
print(f"Detected Outliers: {outliers} (Prices lying outside the 1.5*IQR whiskers)")
```

```python
# Task 4: violinplot() for daily step counts from Google Fit and explain shape
google_fit_steps = np.concatenate([
    np.random.normal(loc=5000, scale=600, size=25), # Weekdays / sedentary
    np.random.normal(loc=11000, scale=1000, size=25) # Active / weekend workouts
])

plt.figure(figsize=(5, 4.5))
sns.violinplot(y=google_fit_steps, color='plum')
plt.title('Daily Step Counts Distribution (Google Fit)')
plt.ylabel('Step Count')
plt.show()

# Explanation:
# The violin plot shows a bimodal shape with two distinct wider bulges (around 5,000 steps and 11,000 steps).
# This reveals two distinct behavioral clusters in user activity: regular sedentary workdays with moderate activity, 
# and highly active workout days or weekend walking routines.

```
