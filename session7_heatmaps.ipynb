# Session 7: Heatmaps & Correlation Analysis
### Topics: Correlation matrix using Pandas, Heatmap with annotations, Diverging color palettes


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Task 1: Create DataFrame with >= 5 columns for food delivery orders and 20 rows, compute correlation
food_df = pd.DataFrame({
    'delivery_time': np.random.randint(15, 60, size=20),
    'order_value': np.random.randint(150, 1200, size=20),
    'tip_amount': np.random.randint(10, 100, size=20),
    'rating': np.round(np.random.uniform(2.5, 5.0, size=20), 1),
    'delivery_distance': np.round(np.random.uniform(1.0, 12.0, size=20), 1)
})

corr_matrix = food_df.corr()
print("Correlation Matrix:")
corr_matrix
```

```python
# Task 2: Heatmap with annotations for food delivery data
plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='Blues', square=True)
plt.title('Food Delivery Features Correlation Heatmap')
plt.show()
```

```python
# Task 3: Diverging color palette (e.g., 'coolwarm' or 'RdYlGn') & identify strongest negative correlation
plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, center=0, square=True)
plt.title('Food Delivery Correlation Heatmap (Diverging Palette: Coolwarm)')
plt.show()

# Find strongest negative correlation
corr_unstacked = corr_matrix.unstack().sort_values()
min_pair = corr_unstacked.index[0]
min_val = corr_unstacked.iloc[0]
print(f"Strongest negative correlation is between {min_pair[0]} and {min_pair[1]}: {min_val:.2f}")
# Explanation: A negative correlation indicates an inverse relationship (e.g., higher delivery distance leading to lower rating or longer delivery times).

```

```python
# Task 4: Mobile phone dataset correlation heatmap
mobile_df = pd.DataFrame({
    'price': [12000, 15000, 22000, 35000, 48000, 65000, 80000, 25000, 18000, 52000],
    'rating': [4.1, 4.0, 4.3, 4.5, 4.6, 4.8, 4.9, 4.2, 4.1, 4.7],
    'RAM_GB': [4, 6, 6, 8, 8, 12, 16, 6, 4, 12],
    'battery_capacity': [4500, 5000, 5000, 4500, 4800, 4400, 4700, 6000, 5000, 5000]
})

mobile_corr = mobile_df.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(mobile_corr, annot=True, fmt='.2f', cmap='vlag', vmin=-1, vmax=1, center=0, square=True)
plt.title('Flipkart Mobile Phones: Correlation Heatmap')
plt.show()
```
