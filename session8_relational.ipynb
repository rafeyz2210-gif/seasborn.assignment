# Session 8: Relational Plots (Scatter, Line, Regressions)
### Topics: sns.relplot(), sns.scatterplot(), sns.lineplot()


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style="whitegrid")
```

```python
# Task 1: scatterplot() for number of followers vs number of posts for 20 Instagram users
insta_users = pd.DataFrame({
    'user': [f'user_{i}' for i in range(1, 21)],
    'posts': np.random.randint(10, 350, size=20),
    'followers': np.random.randint(500, 50000, size=20)
})

plt.figure(figsize=(7, 4))
sns.scatterplot(data=insta_users, x='posts', y='followers', color='magenta', s=70)
plt.title('Instagram Users: Followers vs Number of Posts')
plt.xlabel('Number of Posts')
plt.ylabel('Followers Count')
plt.show()
```

```python
# Task 2: lineplot() for daily food orders on Zomato over 30 days
days_30 = list(range(1, 31))
daily_orders = np.random.randint(120, 350, size=30)

plt.figure(figsize=(9, 4))
sns.lineplot(x=days_30, y=daily_orders, marker='o', color='crimson')
plt.title('Zomato: Daily Food Orders Over a Month')
plt.xlabel('Day of Month')
plt.ylabel('Total Orders Placed')
plt.xticks(range(1, 31, 2))
plt.show()
```

```python
# Task 3: relplot() of movie ratings vs number of reviews across genres
movies_df = pd.DataFrame({
    'movie': [f'Movie {i}' for i in range(1, 16)],
    'rating': [7.2, 8.1, 6.5, 7.8, 8.5, 6.0, 7.4, 8.0, 6.9, 7.7, 8.8, 6.2, 7.1, 8.3, 7.9],
    'reviews': [1200, 2400, 800, 1900, 3100, 600, 1500, 2200, 950, 1800, 3500, 700, 1300, 2900, 2100],
    'genre': ['Action', 'Action', 'Action', 'Action', 'Action',
              'Comedy', 'Comedy', 'Comedy', 'Comedy', 'Comedy',
              'Drama', 'Drama', 'Drama', 'Drama', 'Drama']
})

g = sns.relplot(data=movies_df, x='reviews', y='rating', hue='genre', style='genre', s=100, aspect=1.6)
g.set_axis_labels('Number of Reviews', 'IMDb / Audience Rating')
g.figure.suptitle('Movie Ratings vs Number of Reviews by Genre', y=1.02)
plt.show()
```

```python
# Task 4: regplot() to analyze YouTube hours vs exam scores for 25 students
students_df = pd.DataFrame({
    'youtube_hours': np.random.uniform(0.5, 6.0, size=25)
})
# Inverse relationship with some noise
students_df['exam_score'] = 95 - (students_df['youtube_hours'] * 7.5) + np.random.normal(0, 4, size=25)

plt.figure(figsize=(7, 4.5))
sns.regplot(data=students_df, x='youtube_hours', y='exam_score', color='darkred')
plt.title('YouTube Study/Watch Hours vs Exam Score')
plt.xlabel('Hours Spent on YouTube')
plt.ylabel('Exam Score (%)')
plt.show()

# Interpretation:
# The regression plot demonstrates a clear negative trend (downward sloping line).
# As the number of hours spent on YouTube increases, the resulting exam score decreases on average.

```
