# Session 1: Introduction to Seaborn & Themes
### Topics: Why Seaborn? Figure-level vs axes-level functions, Themes (darkgrid, whitegrid, ticks)


```python
# Task 1: Import libraries and check versions
import seaborn as sns
import matplotlib.pyplot as plt

print(f"Seaborn version: {sns.__version__}")
print(f"Matplotlib version: {plt.matplotlib.__version__}")
```

```python
# Task 2: Daily temperature trend using lineplot()
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

plt.figure(figsize=(8, 4))
sns.lineplot(x=days, y=temps, marker='o', color='crimson')
plt.title('Weekly Temperature Trend')
plt.xlabel('Days of Week')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
# Task 3: Compare Axes-level vs Figure-level functions
# Axes-level: sns.lineplot (drawn directly onto matplotlib Axes)
fig, ax = plt.subplots(figsize=(7, 3.5))
sns.lineplot(x=days, y=temps, ax=ax, marker='s', color='teal')
ax.set_title('Axes-Level: sns.lineplot')
ax.set_xlabel('Day')
ax.set_ylabel('Temp (°C)')
plt.show()

# Figure-level: sns.relplot (manages its own FacetGrid figure)
g = sns.relplot(x=days, y=temps, kind='line', marker='o', color='purple', aspect=1.8, height=3.5)
g.set_axis_labels('Day', 'Temp (°C)')
g.figure.suptitle('Figure-Level: sns.relplot', y=1.03)
plt.show()

# Difference Note:
# 1. Axes-level functions (e.g., sns.lineplot) draw onto a single Matplotlib Axes object and integrate seamlessly with plt.subplots().
# 2. Figure-level functions (e.g., sns.relplot) control the entire figure via FacetGrid, making multi-faceted subplots easy, but they cannot be assigned directly into an existing Matplotlib ax.

```

```python
# Task 4: Apply built-in themes and save separate images
themes = ['darkgrid', 'whitegrid', 'ticks']

for theme in themes:
    sns.set_theme(style=theme)
    plt.figure(figsize=(7, 3.5))
    sns.lineplot(x=days, y=temps, marker='o')
    plt.title(f'Temperature Trend - Theme: {theme}')
    plt.xlabel('Day')
    plt.ylabel('Temp (°C)')
    filename = f'temp_trend_{theme}.png'
    plt.savefig(filename, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Saved {filename}")

# Theme Observations:
# - 'darkgrid': Provides a light-gray background with distinct white gridlines; ideal for data-dense comparison.
# - 'whitegrid': Has a crisp white background with subtle gray gridlines, optimal for cleaner aesthetics.
# - 'ticks': Removes background shading and grids completely, adding tick marks on the axes for a minimalist, print-ready look.

```
