# Session 3: Distribution Plots (Histplot, Kdeplot)
### Topics: sns.histplot(), sns.kdeplot(), Combining hist + KDE


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style="ticks")
```

```python
# Task 1: histplot() of delivery times (in minutes) from Swiggy order dataset (>=30 values)
swiggy_delivery_times = np.random.normal(loc=32, scale=7, size=40).astype(int)

plt.figure(figsize=(7, 4))
sns.histplot(swiggy_delivery_times, bins=10, color='coral', edgecolor='black')
plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Task 2: kdeplot() showing probability density of daily step counts for one month
daily_steps = np.random.normal(loc=7500, scale=1800, size=30).astype(int)

plt.figure(figsize=(7, 4))
sns.kdeplot(daily_steps, fill=True, color='teal', linewidth=2)
plt.title('Daily Step Counts Probability Density (1 Month)')
plt.xlabel('Steps Count')
plt.ylabel('Density')
plt.show()
```

```python
# Task 3: Combine histplot() and kdeplot() for monthly spending amounts on Zomato
monthly_spending = np.random.gamma(shape=5, scale=400, size=50)

plt.figure(figsize=(8, 4))
sns.histplot(monthly_spending, kde=True, color='royalblue', edgecolor='black')
plt.title('Zomato Monthly Spending Distribution (Histogram + KDE)')
plt.xlabel('Spending Amount (₹)')
plt.ylabel('Count')
plt.show()
```

```python
# Task 4: IPL match scores with different bin sizes (5, 10, 20)
ipl_scores = np.random.normal(loc=165, scale=25, size=60).astype(int)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
bins_list = [5, 10, 20]

for i, b in enumerate(bins_list):
    sns.histplot(ipl_scores, bins=b, ax=axes[i], color='purple', edgecolor='black')
    axes[i].set_title(f'IPL Scores (bins = {b})')
    axes[i].set_xlabel('Score')
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

# Written Observation:
# 1. With bins=5: The plot is overly generalized and smoothed; distinct clusters or variance patterns are obscured.
# 2. With bins=10: It provides a balanced visualization where the central peak (around 160-170 runs) and spread are clearly readable.
# 3. With bins=20: The distribution looks noisy and fragmented with multiple small gaps between individual bars due to the limited sample size.

```
