# Session 10: Advanced Seaborn Styling
### Topics: sns.set(), Palettes (colorblind, deep, muted), Context (notebook, talk, poster)


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Task 1: Apply 'poster' context to a barplot of daily Spotify songs streamed
spotify_stream = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'streams': [45000, 48000, 52000, 51000, 68000, 85000, 78000]
})

sns.set_theme(context='poster', style='whitegrid')
plt.figure(figsize=(10, 5))
sns.barplot(data=spotify_stream, x='day', y='streams', palette='Blues_r')
plt.title('Daily Spotify Streams (Poster Context)')
plt.xlabel('Day of Week')
plt.ylabel('Streams')
plt.show()

# Observation: The 'poster' context significantly enlarges all visual elements, including label font sizes, title font size, and bar line weights, ensuring readability from a distance during presentations.

```

```python
# Task 2: countplot showing distribution of food orders using 'colorblind' palette
zomato_types = pd.DataFrame({
    'cuisine_type': ['Indian', 'Chinese', 'Italian', 'Indian', 'Indian', 'Italian', 'Chinese', 'Indian', 'Italian', 'Chinese', 'Indian']
})

sns.set_theme(context='notebook', style='ticks')
plt.figure(figsize=(7, 4))
sns.countplot(data=zomato_types, x='cuisine_type', palette='colorblind')
plt.title('Zomato Order Distribution by Cuisine (Colorblind Safe)')
plt.xlabel('Cuisine Type')
plt.ylabel('Order Count')
plt.show()
```

```python
# Task 3: lineplot of daily step counts (2 weeks) using 'muted' palette and 'notebook' context
two_weeks_steps = pd.DataFrame({
    'day': range(1, 15),
    'steps': [7200, 8100, 6900, 7500, 9200, 11500, 10800, 7400, 7800, 8300, 7900, 9500, 12000, 11000]
})

sns.set_theme(context='notebook', style='whitegrid')
plt.figure(figsize=(8, 4))
sns.lineplot(data=two_weeks_steps, x='day', y='steps', palette='muted', marker='o', color=sns.color_palette("muted")[0])
plt.title('Daily Step Counts - 2 Weeks Trend')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.xticks(range(1, 15))
plt.show()
```

```python
# Task 4: Refactor plot switching palette from 'deep' to 'colorblind' and context from 'talk' to 'poster'
# Demonstration of Talk + Deep:
sns.set_theme(context='talk', palette='deep', style='darkgrid')
plt.figure(figsize=(8, 4))
sns.barplot(data=spotify_stream, x='day', y='streams')
plt.title('Version 1: Talk Context + Deep Palette')
plt.show()

# Demonstration of Poster + Colorblind:
sns.set_theme(context='poster', palette='colorblind', style='darkgrid')
plt.figure(figsize=(10, 5))
sns.barplot(data=spotify_stream, x='day', y='streams')
plt.title('Version 2: Poster Context + Colorblind Palette')
plt.show()

# Brief differences description:
# 1. Palette: 'deep' provides saturated rich tones, whereas 'colorblind' optimizes contrast and hue differentiation to make visual distinctions accessible to individuals with color vision deficiencies.
# 2. Context: 'talk' formats text and markers for laptop/meeting screen sharing, whereas 'poster' scales up font sizes and stroke widths even higher for maximum visibility.

```

```python
# Task 5: Flipkart corporate theme (deep palette, talk context, custom font scale and grid style)
sns.set_theme(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=0.9,
    rc={
        'axes.edgecolor': '#333333',
        'grid.linestyle': '--',
        'grid.alpha': 0.6,
        'axes.titleweight': 'bold'
    }
)

flipkart_cat_sales = pd.DataFrame({
    'category': ['Mobiles', 'Fashion', 'Electronics', 'Home & Kitchen', 'Beauty'],
    'sales_lakhs': [450, 320, 280, 210, 160]
})

plt.figure(figsize=(8, 4.5))
sns.barplot(data=flipkart_cat_sales, x='category', y='sales_lakhs')
plt.title('Flipkart Sales Dashboard: Category-wise Revenue (in Lakhs)')
plt.xlabel('Product Category')
plt.ylabel('Revenue (₹ Lakhs)')
plt.xticks(rotation=15)
plt.show()
```
