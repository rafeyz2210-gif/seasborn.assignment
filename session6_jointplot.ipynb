# Session 6: Jointplot
### Topics: Types: scatter, hexbin, KDE, Relationship analysis: Sales vs Profit


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style="white")
```

```python
# Task 1: Jointplot with kind='scatter' for Flipkart product ratings and prices
flipkart_df = pd.DataFrame({
    'price': np.random.uniform(299, 4999, size=100),
    'rating': np.random.uniform(3.0, 5.0, size=100)
})

g1 = sns.jointplot(data=flipkart_df, x='price', y='rating', kind='scatter', color='royalblue')
g1.fig.suptitle('Flipkart: Price vs Rating (Scatter Jointplot)', y=1.02)
plt.show()
```

```python
# Task 2: Jointplot with kind='hex' for delivery time and order value (Swiggy)
swiggy_orders = pd.DataFrame({
    'delivery_time': np.random.normal(30, 8, 300),
    'order_value': np.random.normal(450, 150, 300)
})

g2 = sns.jointplot(data=swiggy_orders, x='delivery_time', y='order_value', kind='hex', color='darkorange')
g2.fig.suptitle('Swiggy: Delivery Time vs Order Value (Hexbin Jointplot)', y=1.02)
plt.show()
```

```python
# Task 3: Jointplot with kind='kde' for Instagram likes and comments
insta_df = pd.DataFrame({
    'likes': np.random.exponential(scale=500, size=150) + 50,
    'comments': np.random.exponential(scale=35, size=150) + 5
})

g3 = sns.jointplot(data=insta_df, x='likes', y='comments', kind='kde', color='purple', fill=True)
g3.fig.suptitle('Instagram: Likes vs Comments (KDE Jointplot)', y=1.02)
plt.show()
```

```python
# Task 4: Compare scatter, hexbin, and KDE jointplots for Zomato reviews (rating vs number of reviews)
zomato_reviews = pd.DataFrame({
    'rating': np.random.uniform(2.5, 4.9, size=200),
    'reviews_count': np.random.exponential(scale=150, size=200) + 10
})

for kind_type in ['scatter', 'hex', 'kde']:
    g = sns.jointplot(data=zomato_reviews, x='rating', y='reviews_count', kind=kind_type, color='crimson')
    g.fig.suptitle(f'Zomato Reviews: {kind_type.capitalize()} Jointplot', y=1.02)
    plt.show()

# Written Comparison:
# - Scatter: Best for spotting individual outliers and raw data distribution, but suffers from overplotting when points overlap densely.
# - Hexbin: Aggregates points into hexagonal bins with color intensity; ideal for medium-to-large datasets to see local density clearly without visual clutter.
# - KDE: Creates continuous density contours, making it the best choice for identifying underlying probability densities, central modes, and general trends rather than discrete points.

```
