# Session 11: Exporting High-Quality Seaborn Charts
### Topics: Saving with DPI, SVG for presentations, Creating template themes


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

```python
# Task 1: barplot of orders per food category saved as PNG with 300 DPI
food_cat_df = pd.DataFrame({
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [1200, 950, 1450, 800, 600]
})

plt.figure(figsize=(7, 4))
sns.barplot(data=food_cat_df, x='category', y='orders', palette='crest')
plt.title('Food Orders Count by Category')
plt.xlabel('Category')
plt.ylabel('Orders')
plt.savefig('food_orders_300dpi.png', dpi=300, bbox_inches='tight')
plt.show()
print("Saved food_orders_300dpi.png successfully.")
```

```python
# Task 2: line plot of daily step counts exported as SVG for PowerPoint
step_df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'steps': [6500, 7200, 8100, 6900, 8800, 11200, 10500]
})

plt.figure(figsize=(7, 4))
sns.lineplot(data=step_df, x='day', y='steps', marker='o', color='teal', linewidth=2.5)
plt.title('Weekly Daily Step Count')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.savefig('daily_steps.svg', format='svg', bbox_inches='tight')
plt.show()
print("Saved daily_steps.svg successfully.")
```

```python
# Task 3: darkgrid style with chosen palette, saved as PNG at 1200 DPI
sns.set_theme(style='darkgrid', palette='magma')
plt.figure(figsize=(7, 4))
sns.barplot(data=food_cat_df, x='category', y='orders')
plt.title('High-Res Category Orders (1200 DPI)')
plt.xlabel('Category')
plt.ylabel('Orders')
plt.savefig('category_orders_1200dpi.png', dpi=1200, bbox_inches='tight')
plt.show()
print("Saved category_orders_1200dpi.png successfully.")
```

```python
# Task 4: Template function apply_presentation_theme() and apply to top-5 apps pie chart
def apply_presentation_theme(palette_name='Set2'):
    sns.set_theme(
        context='talk',
        style='white',
        palette=palette_name,
        font_scale=0.85,
        rc={
            'axes.edgecolor': '#222222',
            'axes.linewidth': 1.2,
            'font.family': 'sans-serif'
        }
    )

apply_presentation_theme('pastel')

app_data = {
    'apps': ['WhatsApp', 'Instagram', 'YouTube', 'Chrome', 'LinkedIn'],
    'usage_hours': [3.5, 2.5, 2.0, 1.5, 1.0]
}

colors = sns.color_palette('pastel')[:5]
plt.figure(figsize=(6, 6))
plt.pie(app_data['usage_hours'], labels=app_data['apps'], autopct='%1.1f%%', colors=colors, startangle=140)
plt.title('Top 5 Most Used Mobile Apps')
plt.savefig('top_apps_pie.png', dpi=300, bbox_inches='tight')
plt.show()
```
