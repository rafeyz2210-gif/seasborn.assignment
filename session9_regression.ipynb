# Session 9: Regression Plots (lmplot, regplot)
### Topics: Linear regression, Lowess smoothing, Faceted regression: Sales vs Discount


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
sns.set_theme(style="ticks")
```

```python
# Task 1: regplot() for Instagram likes vs comments (>= 10 sample points)
insta_data = pd.DataFrame({
    'likes': [120, 250, 310, 450, 520, 680, 800, 950, 1100, 1300, 1450, 1600],
    'comments': [10, 18, 25, 30, 42, 50, 65, 72, 85, 95, 110, 125]
})

plt.figure(figsize=(7, 4))
sns.regplot(data=insta_data, x='likes', y='comments', color='royalblue')
plt.title('Instagram: Likes vs Comments Linear Regression')
plt.xlabel('Likes')
plt.ylabel('Comments')
plt.show()
```

```python
# Task 2: lmplot() for food order price vs delivery time (Zomato-style, >=15 rows)
zomato_df = pd.DataFrame({
    'order_price': np.random.uniform(150, 900, size=18),
    'delivery_time': np.random.uniform(20, 55, size=18)
})

g = sns.lmplot(data=zomato_df, x='order_price', y='delivery_time', aspect=1.6, height=4)
g.set_axis_labels('Order Price (₹)', 'Delivery Time (min)')
g.figure.suptitle('Zomato: Order Price vs Delivery Time Regression', y=1.03)
plt.show()
```

```python
# Task 3: Apply LOWESS smoothing to regplot() for Spotify songs in playlist vs duration
spotify_df = pd.DataFrame({
    'song_count': np.random.randint(5, 100, size=35)
})
# Non-linear curve
spotify_df['duration_min'] = spotify_df['song_count'] * 3.4 + (spotify_df['song_count']**1.2) * 0.2 + np.random.normal(0, 15, size=35)

plt.figure(figsize=(7, 4.5))
# Linear fit
sns.regplot(data=spotify_df, x='song_count', y='duration_min', label='Linear Fit', scatter=False, color='gray')
# Lowess fit
sns.regplot(data=spotify_df, x='song_count', y='duration_min', lowess=True, color='forestgreen', label='LOWESS Smoothing Fit')
plt.title('Spotify: Song Count vs Playlist Duration (LOWESS Comparison)')
plt.xlabel('Number of Songs')
plt.ylabel('Total Playlist Duration (minutes)')
plt.legend()
plt.show()

# Difference Note:
# The standard linear fit enforces a straight line across all points, whereas lowess=True computes locally weighted polynomial regressions, capturing local curvature and non-linear patterns much better without assuming a global linear relationship.

```

```python
# Task 4: Faceted regression plot (lmplot with col parameter)
study_df = pd.DataFrame({
    'youtube_hours': list(np.random.uniform(1, 6, size=20)) + list(np.random.uniform(1, 6, size=20)),
    'exam_score': list(np.random.uniform(60, 95, size=20)) + list(np.random.uniform(50, 85, size=20)),
    'prep_method': ['Self-study'] * 20 + ['Coaching'] * 20
})

g = sns.lmplot(data=study_df, x='youtube_hours', y='exam_score', col='prep_method', hue='prep_method', aspect=1.2, height=4)
g.set_axis_labels('Hours on YouTube', 'Exam Score (%)')
g.figure.suptitle('YouTube Study Hours vs Exam Score Faceted by Prep Method', y=1.03)
plt.show()
```

```python
# Task 5: Modify given lmplot with custom aesthetics without altering regression logic
flipkart_rev = pd.DataFrame({
    'price': np.random.uniform(500, 5000, size=25),
    'reviews': np.random.uniform(20, 800, size=25)
})

# Custom aesthetics: change marker, scatter color, line color
g = sns.lmplot(
    data=flipkart_rev, 
    x='price', 
    y='reviews',
    scatter_kws={'color': 'darkorange', 'alpha': 0.7, 's': 50, 'marker': '^'},
    line_kws={'color': 'navy', 'linewidth': 2},
    aspect=1.5,
    height=4
)
g.set_axis_labels('Product Price (₹)', 'Number of Reviews')
g.figure.suptitle('Flipkart: Product Price vs Reviews Count', y=1.03)
plt.show()
```
