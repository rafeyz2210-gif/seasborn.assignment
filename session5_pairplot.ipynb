# Session 5: Pairplot (Multivariate Analysis)
### Topics: Pairplot for numerical features, Hue parameter, Scatter + KDE diagonals


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load standard tips dataset
tips = sns.load_dataset('tips')
tips.head()
```

```python
# Task 1: Basic pairplot for all numerical columns
sns.pairplot(tips)
plt.suptitle('Tips Dataset: Basic Pairplot', y=1.02)
plt.show()
```

```python
# Task 2: Pairplot with hue='sex'
sns.pairplot(tips, hue='sex', palette='coolwarm')
plt.suptitle("Tips Dataset: Pairplot by Gender (hue='sex')", y=1.02)
plt.show()
# Observation: Points are colored by male and female customers, showing overlapping distributions with slight variations in maximum spend and tip amounts.

```

```python
# Task 3: Pairplot with hue='smoker' and diag_kind='kde'
sns.pairplot(tips, hue='smoker', diag_kind='kde', palette='Set1')
plt.suptitle("Tips Dataset: Pairplot by Smoker Status with KDE Diagonals", y=1.02)
plt.show()
```

```python
# Task 4: Pairplot with two numerical columns ('total_bill', 'tip') and hue='time'
sns.pairplot(tips, vars=['total_bill', 'tip'], hue='time', palette='Dark2', height=3.5)
plt.suptitle("Tips: total_bill vs tip by Meal Time", y=1.03)
plt.show()
```
